```python
import os, shutil
os.makedirs('/content/train/images', exist_ok=True)
os.makedirs('/content/train/labels', exist_ok=True)

for folder in ['defect_colour', 'defect_missingpiece']:

  for file in os.listdir(f'/content/{folder}/train/labels'):

    with open(f'/content/{folder}/train/labels/{file}', 'r') as f:
      lines = f.readlines()
      #print(f'/content/{folder}/train/labels/{file}', lines)
      if len(lines) == 0:
        continue

    if folder == 'defect_colour':
      lines[0] = '0' + lines[0][1:]
    else:
      lines[0] = '1' + lines[0][1:]

    with open(f'/content/train/labels/{file}', 'w') as f:
      f.writelines(lines)

  for file in os.listdir(f'/content/{folder}/train/images'):
    shutil.copy2(f'/content/{folder}/train/images/{file}', '/content/train/images')
```
